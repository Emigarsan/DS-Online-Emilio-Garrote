#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center" o investigar como funciona bar_label). Los colores de las barras deben tener un 50% de transparencia y ser: azul, gris, rojo, aceituna y marrón.  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Arial", tamaño = 14, peso = "Bold", color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion. Transparencia al 50% y color verde.
4. El boxplot de la altura de las ciudades.

```python
# Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center" o investigar como funciona bar_label). Los colores de las barras deben tener un 50% de transparencia y ser: azul, gris, rojo, aceituna y marrón.  

# Crear una figura con 4 subgráficos en una cuadrícula de 2x2 con tamaño 20x20
fig, ax = plt.subplots(2,2,figsize=(20, 20))

# Ordenar el dataframe por población de mayor a menor
df_cali.sort_values('population_total', ascending=False, inplace=True)

# Obtener las 5 ciudades más pobladas
ciudades = df_cali['city'][0:5]
# Obtener la población en miles de habitantes
poblacion = (df_cali['population_total'][0:5])/1000

# Crear gráfico de barras en la posición [0,0]

# Usar colores especificados con 60% de transparencia (alpha=0.6)
ax[0,0].bar(ciudades, poblacion, color = ['blue', 'gray', 'red', 'olive', 'brown'], alpha = 0.6)

# Obtener el contenedor de barras para etiquetarlas
barras = ax[0,0].containers[0]

# Añadir etiquetas con los valores de población formateados a 2 decimales
ax[0,0].bar_label(barras, labels=[f'{v:.2f}' for v in poblacion], padding=3)

# Establecer límite máximo del eje Y en 5000
ax[0,0].set_ylim(0,5000)

# Añadir títulos y etiquetas de ejes
ax[0,0].set_title('Ciudades más pobladas de California')
ax[0,0].set_xlabel('Ciudades')
ax[0,0].set_ylabel('Habitantes (en miles)');
```

```python
# Histograma del área total ocupada por las ciudades:
# - 50 bins
# - color "steelblue"
# - tipografía Arial, tamaño 14, negrita

# Dibuja el histograma en el subplot superior derecho (fila 0, columna 1)
ax[0,1].hist(
    df_cali['area_total_km2'],  # Columna con el área total en km²
    bins=50,                    # Número de barras del histograma
    color='steelblue',          # Color
    alpha=0.7                   # Transparencia (70%)
)

# Define el estilo de fuente común para título y ejes
font_dict = {
    'family': 'Arial',
    'size': 14,
    'weight': 'bold'
}

# Configura título y etiquetas de ejes con la fuente indicada
ax[0,1].set_title('Histograma del área ocupada por las ciudades', fontdict=font_dict)
ax[0,1].set_xlabel('Superficie ($km^{2}$)', fontdict=font_dict)
ax[0,1].set_ylabel('Número de ciudades', fontdict=font_dict)

# Añade una leyenda para identificar el histograma
ax[0,1].legend(['Hist Area'], loc='upper right')

# Muestra la figura completa (los 4 subgráficos)
display(fig)
```

```python
# El scatter de área de agua de una ciudad frente a su población (transparencia 50%, color verde)
area_agua = df_cali['area_water_km2']                       # Serie con el área de agua por ciudad
poblacion_total = df_cali['population_total']/1000          # Población en miles de habitantes
ax[1,0].scatter(area_agua, poblacion_total, color='green', alpha=0.5)

# Títulos y etiquetas del gráfico
ax[1,0].set_title('Población por kilometro cuadrado de agua')
ax[1,0].set_xlabel('Área de agua en la ciudad ($km^{2}$)', fontsize=12)
ax[1,0].set_ylabel('Población (miles habitantes)', fontsize=12)

# Calcular outliers como tuplas (area_water_km2, población en miles)
condicion_outliers = (df_cali['area_water_km2'] > 80) & (df_cali['population_total'] > 1000)
x_outliers = df_cali.loc[condicion_outliers, 'area_water_km2']
y_outliers = df_cali.loc[condicion_outliers, 'population_total'] / 1000
outliers = list(zip(x_outliers, y_outliers))

# Posición para el texto central que etiqueta los outliers (un único texto con flechas)
text_x = 70
text_y = 3000
ax[1,0].text(text_x, text_y, 'Outliers', ha='center', va='center', fontsize=12)

# Dibujar flechas desde el texto central hacia cada punto outlier (ahora tuplas x,y)
for x_out, y_out in outliers:
    ax[1,0].annotate(
        '',
        xy=(x_out, y_out),
        xytext=(text_x, text_y),
        arrowprops=dict(facecolor='black', arrowstyle='->', shrinkA=10, shrinkB=5)
    )

# Mostrar la figura completa con los cuatro subgráficos
display(fig)
```

```python
# Boxplot de la altura (elevation_m) de las ciudades
ax[1,1].boxplot(
    df_cali['elevation_m'],                 # Datos: elevación en metros
    vert=False,                             # Orientación horizontal
    patch_artist=True,                      # Permite colorear el interior de la caja
    boxprops=dict(facecolor='red', color='black'),  # Estilo de la caja
    medianprops=dict(color='black'),        # Estilo de la línea de la mediana
)

# Título y etiqueta del eje X
ax[1,1].set_title('Diagrama de caja de la Altitud de las ciudades de California')
ax[1,1].set_xlabel('Altitud (m)', fontsize=12)

# Oculta la escala del eje Y (no aporta información en un boxplot horizontal con una sola serie)
ax[1,1].set_yticks([])

# Muestra la figura completa con los 4 subgráficos
display(fig)
```
